```python
import yfinance as yf
import pandas as pd
import numpy as np

# STEP 1: List the 5 tickers we want to analyze
# BTC-USD = Bitcoin | DX-Y.NYB = US Dollar Index | EURUSD=X = Euro/Dollar | AAPL = Apple | TSLA = Tesla
tickers = ['BTC-USD', 'DX-Y.NYB', 'EURUSD=X', 'AAPL', 'TSLA']

# STEP 2: Download the last 1 year of daily closing prices from Yahoo Finance
prices = yf.download(tickers, period='1y', progress=False)['Close']

# Fill in weekend gaps (since Bitcoin trades on weekends but stocks/forex don't)
prices = prices.ffill().dropna()

# STEP 3: Calculate Daily % Returns (How much % each asset moved each day)
daily_returns = prices.pct_change().dropna()

# STEP 4: Calculate Risk Metrics using 3 simple formulas
# A. Average Daily Volatility (%) -> Standard Deviation of daily returns
daily_volatility_pct = daily_returns.std() * 100

# B. Annualized Volatility (%) -> Daily volatility multiplied by square root of 252 trading days
annual_volatility_pct = daily_volatility_pct * np.sqrt(252)

# C. Maximum Drawdown (%) -> The worst percentage drop from a peak price to a bottom price
highest_peak_price = prices.cummax()
drop_from_peak_pct = ((prices - highest_peak_price) / highest_peak_price) * 100
max_drawdown_pct = drop_from_peak_pct.min()

# STEP 5: Put the results into one clean table
risk_table = pd.DataFrame({
    'Latest Price': prices.iloc[-1].round(4),
    'Avg Daily Volatility (%)': daily_volatility_pct.round(2),
    'Annual Volatility (%)': annual_volatility_pct.round(2),
    'Max Drawdown (%)': max_drawdown_pct.round(2)
})

print("=== 1-YEAR MULTI-ASSET VOLATILITY & DRAWDOWN TABLE ===")
display(risk_table)
```
